```python
# Databricks notebook source

from pyspark.sql.functions import col,round,lit

df = spark.table("dev_prct.dev_asset_prct.landing_asset_check")


# DBTITLE 1,Transform / aggregate
transformed_df = df.select(
    col("file_path").alias("File_Path")
).withColumn("location",lit('chennai'))

transformed_df.display()

# COMMAND ----------

# DBTITLE 1,Write to Delta (Silver layer)
transformed_df.write.format("delta") \
    .mode("overwrite") \
    .option("overwriteSchema", "true") \
    .saveAsTable("dev_prct.dev_asset_prct.landing_asset_edited")

print("Transformation complete: silver_population_data")
```
